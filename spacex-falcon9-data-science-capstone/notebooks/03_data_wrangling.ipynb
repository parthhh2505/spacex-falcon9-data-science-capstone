# 03 — Data Wrangling

```python
import pandas as pd
import numpy as np

DATA_URL = "https://raw.githubusercontent.com/adgsenpai/IBM-DataScience-SpaceX-Capstone/main/dataset_part_2.csv"
df = pd.read_csv(DATA_URL)
print(df.shape)
df.head()
```

```python
print(df.isna().sum().sort_values(ascending=False).head(10))

df["PayloadMass"] = df["PayloadMass"].fillna(df["PayloadMass"].mean())
df["Class"] = pd.to_numeric(df["Class"], errors="coerce").fillna(0).astype(int)
df["Date"] = pd.to_datetime(df["Date"])
df["Year"] = df["Date"].dt.year

df.to_csv("../data/spacex_cleaned.csv", index=False)
print("Saved cleaned dataset:", df.shape)
```

Key preparation: missing-value handling, date conversion, year extraction, binary landing class, categorical encoding and feature preparation.
