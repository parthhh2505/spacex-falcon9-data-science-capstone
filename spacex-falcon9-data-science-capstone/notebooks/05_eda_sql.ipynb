# 05 — EDA with SQL

```python
import pandas as pd
import sqlite3

df = pd.read_csv("../data/spacex_cleaned.csv")
conn = sqlite3.connect("../data/spacex.db")
df.to_sql("SPACEXTBL", conn, if_exists="replace", index=False)
```

```python
queries = {
    "distinct_sites": "SELECT DISTINCT LaunchSite FROM SPACEXTBL;",
    "cca_records": "SELECT * FROM SPACEXTBL WHERE LaunchSite LIKE 'CCA%' LIMIT 5;",
    "avg_payload": "SELECT AVG(PayloadMass) AS avg_payload FROM SPACEXTBL WHERE BoosterVersion LIKE '%v1.1%';",
    "first_success": "SELECT MIN(Date) AS first_successful_landing FROM SPACEXTBL WHERE Class=1 AND LandingPad IS NOT NULL;",
    "rank_sites": "SELECT LaunchSite, COUNT(*) AS launches FROM SPACEXTBL GROUP BY LaunchSite ORDER BY launches DESC;"
}

for name, query in queries.items():
    print("\n---", name, "---")
    print(pd.read_sql_query(query, conn).to_string(index=False))
```

```python
query = '''
SELECT FlightNumber, PayloadMass, Orbit, LaunchSite, Class
FROM SPACEXTBL
WHERE PayloadMass BETWEEN 2000 AND 5000
  AND Class = 1
ORDER BY PayloadMass;
'''
print(pd.read_sql_query(query, conn).to_string(index=False))
```
