# 01 — SpaceX API Data Collection

```python
import requests
import pandas as pd

url = "https://api.spacexdata.com/v4/launches/past"
response = requests.get(url, timeout=30)
response.raise_for_status()
launches = response.json()

df = pd.json_normalize(launches)
print("Raw launch records:", len(df))
print("Columns:", len(df.columns))
df.head()
```

```python
# Save the raw API response for inspection.
df.to_csv("../data/api_raw_launches.csv", index=False)
print("Saved data/api_raw_launches.csv")
```
