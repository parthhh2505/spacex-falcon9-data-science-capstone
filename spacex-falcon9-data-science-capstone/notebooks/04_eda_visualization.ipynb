# 04 — EDA & Data Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("https://raw.githubusercontent.com/adgsenpai/IBM-DataScience-SpaceX-Capstone/main/dataset_part_2.csv")
df["PayloadMass"] = df["PayloadMass"].fillna(df["PayloadMass"].mean())
df["Date"] = pd.to_datetime(df["Date"])
df["Year"] = df["Date"].dt.year
```

```python
plt.figure(figsize=(9,5))
sns.countplot(data=df, x="LaunchSite")
plt.xticks(rotation=25)
plt.title("Falcon 9 Launches by Site")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x="FlightNumber", y="PayloadMass", hue="LaunchSite")
plt.title("Payload Mass Across Flight Sequence")
plt.tight_layout()
plt.show()
```

```python
orbit_success = df.groupby("Orbit")["Class"].mean().sort_values(ascending=False)
plt.figure(figsize=(10,5))
orbit_success.plot(kind="bar")
plt.ylabel("Landing success rate")
plt.title("Landing Success Rate by Orbit")
plt.tight_layout()
plt.show()
```

```python
yearly = df.groupby("Year")["Class"].mean()
plt.figure(figsize=(10,5))
yearly.plot(marker="o")
plt.ylabel("Landing success rate")
plt.title("Landing Success Rate by Year")
plt.grid(True, alpha=.25)
plt.tight_layout()
plt.show()
```

Required EDA coverage: launch sites, payloads, orbit, flight sequence and yearly landing-success trend.
